```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
airbnb = pd.read_csv('/Users/davidmarip/Downloads/DS2002/airbnb_dataset.csv')
```

```python
print(airbnb.head())
```

```python
#Price distribution
plt.figure(figsize=(10, 6))
plt.hist(airbnb['Price'], bins=30, color='skyblue', edgecolor='black')
plt.title('Distribution of Airbnb Nightly Rates')
plt.xlabel('Price ($)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#Locations with pricing
plt.figure(figsize=(10, 6))
plt.scatter(airbnb['Longitude'], airbnb['Latitude'], c=airbnb['Price'], cmap='viridis', alpha=0.7)
plt.colorbar(label='Price ($)')
plt.title('Geographic Distribution of Airbnb Listings')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

```python
#Made it interactive
fig = px.scatter(
    airbnb,
    x='Longitude',
    y='Latitude',
    color='Price',
    title='Airbnb Listings with Prices (Interactive)',
    labels={'Longitude': 'Longitude', 'Latitude': 'Latitude', 'Price': 'Nightly Rate ($)'},
    hover_data=['Price']
)
fig.show()
```

```python
fig = px.box(
    airbnb,
    y='Price',
    title='Box Plot of Nightly Rates',
    labels={'Price': 'Nightly Rate ($)'},
)
fig.show()
```
